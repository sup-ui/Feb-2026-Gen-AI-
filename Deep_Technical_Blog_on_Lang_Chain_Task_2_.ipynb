**LLMs and Chat Models**

In this section, we look at how Large Language Models (LLMs) can be used within LangChain to generate intelligent responses. LLMs are powerful AI models capable of understanding natural language and producing meaningful text based on the input they receive.

LangChain provides support for Chat Models, which work using a structured message format rather than plain text. This format organizes communication into different roles:

System Message – Defines the behavior or instructions for the AI model.
User (Human) Message – Represents the input or question provided by the user.
AI Message – The response generated by the model.

Using this structured interaction makes it easier to design real-world applications such as chatbots, virtual assistants, and AI-powered tools. It also allows developers to control how the model behaves and how conversations flow within an application.

**Prompt Templates**

In many simple examples, prompts are created using Python f-strings. While this approach works for small experiments, it becomes difficult to manage when building larger applications. Hardcoding prompts can make the code messy and harder to maintain.

Prompt Templates in LangChain help solve this issue by providing a structured way to design prompts. Instead of writing the entire prompt every time, developers can create a reusable template and later insert the required inputs dynamically.

Prompt Templates provide several advantages:

* They make prompts reusable across different tasks
* They separate the prompt structure from the user input
* They allow dynamic values to be inserted easily

By using templates, developers can design cleaner and more scalable applications where prompts can be reused and modified without rewriting the entire code.

**Multi-Input Prompt System**

```python
from langchain_core.prompts import PromptTemplate

# Create multi-input template
template = "Explain {topic} for {audience} in a {tone} tone."

prompt = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template
)

# Test Case 1
output1 = prompt.format(
    topic="Artificial Intelligence",
    audience="beginners",
    tone="friendly"
)

print("Test Case 1:")
print(output1)

# Test Case 2
output2 = prompt.format(
    topic="Python",
    audience="kids",
    tone="fun"
)

print("\nTest Case 2:")
print(output2)

# Test Case 3
output3 = prompt.format(
    topic="Deep Learning",
    audience="engineers",
    tone="technical"
)

print("\nTest Case 3:")
print(output3)
```

**Prompt Variations Engine**

```python
from langchain_core.prompts import PromptTemplate

# Teaching Template
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step."
)

# Interview Template
interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}."
)

# Storytelling Template
story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story."
)

# Input
topic = "Machine Learning"

# Outputs
print("Teaching Style:")
print(teaching_template.format(topic=topic))

print("\nInterview Style:")
print(interview_template.format(topic=topic))

print("\nStorytelling Style:")
print(story_template.format(topic=topic))
```

**ChatPromptTemplate (Roles System)**

```python
from langchain_core.prompts import ChatPromptTemplate

# Create chat template
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a {role}."),
    ("human", "Explain {topic} in simple terms.")
])

# Test Case 1
messages1 = chat_template.format_messages(
    role="teacher",
    topic="Neural Networks"
)

print("Test Case 1:")
for msg in messages1:
    print(msg)

# Test Case 2
messages2 = chat_template.format_messages(
    role="motivator",
    topic="Success"
)

print("\nTest Case 2:")
for msg in messages2:
    print(msg)
```

**Input Validation + Prompt Generator**

```python
from langchain_core.prompts import PromptTemplate

# Step 1: Validation Function
def validate_inputs(audience, tone):

    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        print("Invalid audience! Setting default: beginner")
        audience = "beginner"

    if tone not in valid_tone:
        print("Invalid tone! Setting default: formal")
        tone = "formal"

    return audience, tone


# Step 2: Prompt Generator Function
def generate_prompt(topic, audience, tone, style):

    audience, tone = validate_inputs(audience, tone)

    # Different styles
    if style == "teaching":
        template = "Explain {topic} for {audience} in a {tone} teaching style."

    elif style == "interview":
        template = "Ask interview questions about {topic} for {audience} in a {tone} tone."

    elif style == "storytelling":
        template = "Explain {topic} as a story for {audience} in a {tone} tone."

    else:
        template = "Explain {topic} for {audience} in a {tone} tone."

    prompt = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template=template
    )

    return prompt.format(
        topic=topic,
        audience=audience,
        tone=tone
    )


# 🔹 Test Case 1
print("Test Case 1:")
print(generate_prompt("AI", "beginner", "fun", "storytelling"))

# 🔹 Test Case 2 (Invalid input test)
print("\nTest Case 2:")
print(generate_prompt("Python", "kids", "cool", "teaching"))
```

**Template Reusability Test**

```python
from langchain_core.prompts import PromptTemplate

# One reusable template
template = "Explain {topic} for {audience} in a {tone} tone."

prompt = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template=template
)

# 5 different test cases
test_cases = [
    {"topic": "AI", "audience": "beginner", "tone": "fun"},
    {"topic": "Python", "audience": "intermediate", "tone": "casual"},
    {"topic": "Deep Learning", "audience": "expert", "tone": "formal"},
    {"topic": "Data Science", "audience": "beginner", "tone": "formal"},
    {"topic": "Machine Learning", "audience": "intermediate", "tone": "fun"},
]

# Run all test cases
for i, case in enumerate(test_cases, 1):
    print(f"Test Case {i}:")
    print(prompt.format(**case))
    print()
```
